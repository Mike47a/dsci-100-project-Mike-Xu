```python
import altair as alt
```

```python
win_distribution = alt.Chart(tidy_data).mark_bar().encode(
    x=alt.X('reg_wins_pct:Q',bin=True).title('Regular Season Win Percentage (%)'),
    y=alt.Y('count():Q').title('Number of Team Records')
).properties(title='Distribution of Regular Season Win Percentage')
win_distribution
```

```python
home_counts = (tidy_data.groupby('first_game_at_home')
               .size()
              .reset_index(name='count'))
home_counts = home_counts.assign(proportion = home_counts['count']/home_counts['count'].sum())
home_proportion_plot = alt.Chart(home_counts).mark_bar().encode(
    x=alt.X("first_game_at_home:N").title("First Game at Home (Y = Yes, N = No)"),
    y=alt.Y("proportion:Q").title("Proportion of Team Records"),
    color=alt.Color('first_game_at_home:N').title('First Game At Home')
)
home_proportion_plot
```

Visualization 1: Teams with at least four tournament wins mostly had regular season win percentages above about 65%, while teams with six wins had percentages of about 70% or higher. However, many teams with win percentages between 80% and 100% still had zero tournament wins. This suggests that regular season win percentage alone may not be enough to predict tournament wins, so my analysis also considers seed number and whether the first game was at home.

```python
win_relationship = (
    alt.Chart(tidy_data)
    .mark_point(opacity=0.3,size=20)
    .encode(
        x=alt.X("reg_wins_pct:Q")
        .title("Regular Season Win Percentage (%)").
        scale(zero=False),
        y=alt.Y("tourney_wins:Q").title("Tournament Wins")
    )
    .properties(title="Regular Season Win Percentage and Tournament Wins")
)

win_relationship
```

Visualization 2: A smaller seed number means a higher ranking. Only teams with seed numbers 1–3 won six tournament games. Bigger circles mean more team records. Teams with smaller seed numbers generally won more games. This pattern is clearer than the pattern in Visualization 1. However, teams seeded 1 had outcomes ranging from zero to six wins, so seed number alone cannot determine tournament wins. This supports considering seed number alongside regular season win percentage and first-game home status in my prediction model.

```python
seed_relationship = alt.Chart(tidy_data).mark_circle(opacity=0.8, size=20).encode(
        x=alt.X("seed").title("Seed Number"),
        y=alt.Y("tourney_wins").title("Tournament Wins"),
        size=alt.Size("count():Q").title("Number of Teams")
    ).properties(title="Seed Number and Tournament Wins")
seed_relationship
```

------


Visualization 3: Teams whose first game was at home averaged 1.95 tournament wins, compared with 0.56 for teams whose first game was not at home. This suggests that home status could help predict tournament wins. However, higher-ranked teams often start the tournament at home, so this difference may partly reflect seed number, as shown in Visualization 2. Home status and seed number may therefore provide overlapping information. I will consider both variables together with regular season win percentage in my model.

```python
home_relationship = alt.Chart(tidy_data).mark_bar().encode(
        x=alt.X("first_game_at_home:N").title(
            "First Game at Home (Y = Yes, N = No)"),
        y=alt.Y("mean(tourney_wins):Q").title("Mean Tournament Wins"),
        color=alt.Color("first_game_at_home:N").title("First Game at Home")
    ).properties(title="Mean Tournament Wins by First Game Home Status")
home_relationship
```

I will use k-nearest neighbours (kNN) regression to predict tournament wins using seed number, regular season win percentage, and first-game home status. Tournament wins is numeric, so regression is suitable. Visualizations 1 and 2 suggest curved patterns, especially for seed number; kNN can follow these patterns without assuming a straight-line relationship.

Before modeling, I will remove records with seed = 0 because zero is not a regular seed ranking. I will randomly split the remaining data into 75% training and 25% test data, using a fixed random seed. I will standardize the seed number and regular season win percentage and convert home status to Y = 1 and N = 0.

I will use five-fold cross-validation on the training data to compare different values of k. Scaling will be fitted separately within each training fold. I will choose the k with the lowest average root mean squared prediction error (RMSPE), then fit the final model on all training data. No separate validation set is needed because cross-validation already provides validation data in each round.

I will evaluate the final model once on the test data using RMSPE. It measures prediction error in wins and gives larger errors a greater penalty. Lower RMSPE means better predictions. One limitation is that seed number and home status provide overlapping information. This may give team strength too much weight when kNN measures how similar teams are.
